# 01. Data Audit and Cleaning

This notebook reconstructs the raw forest-fire file into a valid tabular dataset, repairs the known malformed observation, standardizes data types and categorical text, validates the cleaned data, and saves the result for later analysis.

Feature engineering is intentionally left for the next notebook.


## 1. Imports and Data Path


```python
from pathlib import Path

import pandas as pd
```

```python
data_path = Path("forest_fires_dataset.csv")
data_path.exists()
```

## 2. Inspect the Raw File Structure

The raw CSV contains region labels, repeated headers, and blank structural rows, so it cannot be read directly as one ordinary table.


```python
with data_path.open("r", encoding="utf-8") as file:
    for line_number in range(15):
        line = file.readline()
        print(line_number + 1, repr(line))
```

```python
df_raw = pd.read_csv(
    data_path,
    header=None,
    dtype=str
)

print("Raw shape:", df_raw.shape)
df_raw.head(10)
```

## 3. Reconstruct the Observation Table

The two region marker rows are used to create a `Region` column.  
Rows with a numeric first field are then retained as actual observations.


```python
region_names = ["Cordillera", "Hudson Bay"]

is_region_row = df_raw[0].str.strip().isin(region_names)

df_raw.loc[is_region_row]
```

```python
# Assign each region marker to its following observations
df_raw["Region"] = df_raw[0].where(is_region_row)
df_raw["Region"] = df_raw["Region"].ffill()
```

```python
# Observation rows have a numeric day value in the first column
day_numeric = pd.to_numeric(
    df_raw[0].str.strip(),
    errors="coerce"
)

is_observation = day_numeric.notna()

df_clean = df_raw.loc[is_observation].copy()

print("Observation rows:", df_clean.shape[0])
```

```python
# The true column names are stored in the first repeated header row
column_names = (
    df_raw.loc[1, list(range(14))]
    .str.strip()
    .tolist()
)

df_clean.columns = column_names + ["Region"]
df_clean.columns = df_clean.columns.str.strip()

df_clean.reset_index(drop=True, inplace=True)

df_clean.head()
```

## 4. Repair the Known Malformed Observation

One raw row contains a missing comma, which shifts the final values to the right.  
The verified values are corrected explicitly.


```python
df_clean[df_clean["Classes"].isna()]
```

```python
df_clean.loc[165]
```

```python
# Correct the known malformed row
df_clean.loc[
    165,
    ["DC", "ISI", "BUI", "FWI", "Classes"]
] = ["14.6", "9", "12.5", "10.4", "fire"]

df_clean.loc[165]
```

## 5. Standardize Text and Data Types


```python
# Clean categorical text
df_clean["Classes"] = df_clean["Classes"].str.strip()
df_clean["Region"] = df_clean["Region"].str.strip()
```

```python
numeric_columns = [
    "day",
    "month",
    "year",
    "Temperature",
    "RH",
    "Ws",
    "Rain",
    "FFMC",
    "DMC",
    "DC",
    "ISI",
    "BUI",
    "FWI"
]

df_clean[numeric_columns] = df_clean[numeric_columns].apply(
    pd.to_numeric,
    errors="raise"
)

df_clean.dtypes
```

## 6. Validate the Cleaned Dataset

These checks confirm the expected number of observations, absence of missing values and duplicates, valid categorical values, and valid calendar dates.

The date conversion here is only a validation check; no new `Date` feature is saved during data cleaning.


```python
print("Shape:", df_clean.shape)
print("\nMissing values:")
print(df_clean.isna().sum())
```

```python
print("Classes:")
print(df_clean["Classes"].value_counts(dropna=False))

print("\nRegions:")
print(df_clean["Region"].value_counts(dropna=False))
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# Validate that day/month/year form real calendar dates
date_check = pd.to_datetime(
    df_clean[["year", "month", "day"]],
    errors="raise"
)

print("Earliest date:", date_check.min())
print("Latest date:", date_check.max())
```

```python
# Final structural checks
assert df_clean.shape == (244, 15)
assert df_clean.isna().sum().sum() == 0
assert df_clean.duplicated().sum() == 0
assert set(df_clean["Classes"].unique()) == {"fire", "not fire"}
assert set(df_clean["Region"].unique()) == {"Cordillera", "Hudson Bay"}

df_clean.info()
```

## 7. Save the Cleaned Dataset

The cleaned dataset retains the original `day`, `month`, and `year` variables.  
Any redesign of these variables belongs to feature engineering in Notebook 02.


```python
df_clean.to_csv("forest_fires_clean.csv", index=False)
```
