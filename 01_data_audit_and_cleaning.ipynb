```python
from pathlib import Path

import numpy as np
import pandas as pd 
```

```python
data_path = Path("forest_fires_dataset.csv")
data_path.exists()
```

```python
with data_path.open("r", encoding="utf-8") as file:
    for line_number in range(15):
        line = file.readline()
        print(line_number + 1, repr(line))
```

```python
df_raw = pd.read_csv(
    data_path,
    header=None,
    dtype=str
)

df_raw.head(10)
```

```python
df_raw.shape
```

```python
region_names = ["Cordillera", "Hudson Bay"]

is_region_row = (
    df_raw[0].str.strip().isin(region_names)
)

df_raw.loc[is_region_row]
```

```python
# Add a new column for the region and fill it down the rows
df_raw["Region"] = df_raw[0].where(is_region_row)
df_raw["Region"] = df_raw["Region"].ffill()

df_raw.head(10)
```

```python
day_numeric = pd.to_numeric(
    df_raw[0].str.strip(),
    errors="coerce"
)

# a clean observation has a numeric day value, so we can use that to filter out the non-observation rows
is_observation = day_numeric.notna()
df_clean = df_raw.loc[is_observation].copy()
df_clean.shape
```

```python
# The actual column names are stored in row 1 of df_raw

column_names = (
    df_raw.loc[1, list(range(14))]
    .str.strip()
    .tolist()
)

df_clean.columns = column_names + ["Region"]

df_clean.head()
```

```python
df_raw.info()
```

```python
df_clean.reset_index(drop=True, inplace=True)
df_clean.info()
```

```python
df_clean[df_clean["Classes"].isna()]
```

```python
df_clean.loc[165]
```

```python
# row is shifted due to a missing comma

df_clean.loc[
    165,
    ["DC", "ISI", "BUI", "FWI", "Classes"]
] = ["14.6", "9", "12.5", "10.4", "fire"]

df_clean.loc[165]
```

```python
df_clean.info()
```

```python
numeric_columns = [
    "day",
    "month",
    "year",
    "Temperature",
    "RH",
    "Ws",
    "Rain",
    "FFMC",
    "DMC",
    "DC",
    "ISI",
    "BUI",
    "FWI"
]

df_clean[numeric_columns] = df_clean[numeric_columns].apply(
    pd.to_numeric,
    errors="raise"
)

df_clean.dtypes
```

```python
df_clean["Date"] = pd.to_datetime(
    df_clean[["year", "month", "day"]]
)

df_clean.info()
```

```python
# Check categorical values
print(df_clean["Classes"].value_counts(dropna=False))
print(df_clean["Region"].value_counts(dropna=False))
```

```python
# strip str columns
df_clean["Classes"] = df_clean["Classes"].str.strip()
df_clean["Region"] = df_clean["Region"].str.strip()

# strip column names
df_clean.columns = df_clean.columns.str.strip()

# verify the categories
print(df_clean["Classes"].unique())
print(df_clean["Region"].unique())
```

```python
# check for duplicates
df_clean.duplicated().sum()
```

```python
df_clean.to_csv("forest_fires_clean.csv", index=False)
```
