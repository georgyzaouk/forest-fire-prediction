## 1. Setup


```python
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

## 2. Load Engineered Train and Test Sets


```python
train_df = pd.read_csv("forest_fires_train_engineered.csv")
test_df = pd.read_csv("forest_fires_test_engineered.csv")
```

## 3. Define Predictors and Target


```python
X_train = train_df.drop(columns="Classes")
y_train = train_df["Classes"]

X_test = test_df.drop(columns="Classes")
y_test = test_df["Classes"]

# Encode the target so that fire is the positive class
target_mapping = {
    "not fire": 0,
    "fire": 1
}

y_train = y_train.map(target_mapping)
y_test = y_test.map(target_mapping)
```

## 4. Preprocessing

Preprocessing is defined here and later attached directly to model pipelines so that every transformation is fitted only on the training folds during cross-validation.


### 4.1 Feature Groups


```python
numeric_features = [
    "Temperature",
    "RH",
    "Ws",
    "Rain",
    "FFMC",
    "DMC",
    "DC",
    "ISI",
    "BUI",
    "FWI",
    "DayOfYear"
]

categorical_features = ["Region"]
```

### 4.2 Preprocessor for Models That Need Scaling


```python
scaled_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                drop="if_binary",
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)
```

### 4.3 Preprocessor for Tree-Based Models


```python
tree_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                drop="if_binary",
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)
```
